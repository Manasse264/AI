# AI Assistant for Question Answering

## Problem Domain
Natural Language Processing (NLP) / Question Answering

## AI Approach
LLM-based application using a pretrained language model.

## Main Functionality
The application accepts a natural language question and generates an answer by processing context from a knowledge base using a pretrained language model.

## Input
Natural language text query (question).

## Processing
Question tokenization, document/context association, and Transformer inference.

## Output
Extracted string answer with an associated confidence score.

# System Architecture

User (Input Query)
       │
       ▼
Streamlit / Jupyter Interface
       │
       ▼
Data Context & Prompt Preprocessing
       │
       ▼
Pretrained Model (distilbert-base-cased-distilled-squad)
       │
       ▼
Answer Extraction & Confidence Scoring
       │
       ▼
Output Display (User Response)

```python
from transformers import pipeline

print("Loading question-answering model...")
qa_model = pipeline(
    "question-answering",
    model="distilbert-base-cased-distilled-squad"
)
print("Question-answering model loaded successfully.")
```

```python
sample_context = """
Artificial intelligence (AI) is a field of computer science
that focuses on creating systems capable of performing tasks
that normally require human intelligence, such as learning,
reasoning, understanding language, and recognizing patterns.
"""
sample_question = "What is artificial intelligence?"

result = qa_model(
    question=sample_question,
    context=sample_context
)

print("Question:", sample_question)
print("Answer:", result["answer"])
print("Confidence Score:", round(result["score"], 4))
```

```python
import json

# 1. Acquire Data (Domain Knowledge Base)
knowledge_base = [
    {
        "id": 1,
        "category": "AI Concepts",
        "title": "Artificial Intelligence Overview",
        "content": "Artificial intelligence (AI) is a branch of computer science focused on building smart machines capable of performing tasks that typically require human intelligence, such as visual perception, speech recognition, decision-making, and language translation."
    },
    {
        "id": 2,
        "category": "Machine Learning",
        "title": "Supervised Learning",
        "content": "Supervised learning is an approach to machine learning where a computer algorithm is trained on input data that has been labeled for a particular output. The model learns to predict outputs for unlabelled data."
    },
    {
        "id": 3,
        "category": "NLP",
        "title": "Natural Language Processing",
        "content": "Natural Language Processing (NLP) is a subfield of computer science and AI concerned with giving computers the ability to understand text and spoken words in much the same way human beings can."
    }
]

# Save dataset locally to satisfy data source acquisition requirement
with open("dataset.json", "w") as f:
    json.dump(knowledge_base, f, indent=4)

# 2. Data Preprocessing
processed_docs = []
for item in knowledge_base:
    clean_text = item["content"].strip()
    processed_docs.append(clean_text)

full_context = " ".join(processed_docs)

print("Data acquisition & preprocessing complete.")
print(f"Total processed context length: {len(full_context)} characters.")
```

```python
def ask_assistant(question, context, top_k=1):
    """
    Function Parameters:
    - question (str): The user input query.
    - context (str): Knowledge base context text.
    - top_k (int): Number of top candidate answers returned by the model.
    """
    if not question.strip():
        return "Please enter a valid question."
    
    response = qa_model(
        question=question,
        context=context,
        top_k=top_k
    )
    return response

# Test engineered query pipeline
test_q = "What does supervised learning use for training?"
ans = ask_assistant(test_q, full_context)

print("Query:", test_q)
print("Answer:", ans["answer"])
print("Score:", round(ans["score"], 4))
```

```python
# Test suite for model evaluation
test_eval_set = [
    {"q": "What is artificial intelligence?", "expected_key": "branch of computer science"},
    {"q": "What does supervised learning use for training?", "expected_key": "labeled"},
    {"q": "What ability does NLP give computers?", "expected_key": "understand text"}
]

correct_count = 0
total = len(test_eval_set)

print("--- EVALUATION REPORT ---")
for idx, test in enumerate(test_eval_set, 1):
    res = ask_assistant(test["q"], full_context)
    predicted = res["answer"]
    score = res["score"]
    
    # Verify expected key term exists in output
    is_correct = test["expected_key"].lower() in predicted.lower()
    if is_correct:
        correct_count += 1
        
    print(f"Test {idx}: {test['q']}")
    print(f"  Prediction: '{predicted}' (Score: {score:.4f})")
    print(f"  Status: {'PASS' if is_correct else 'FAIL'}\n")

accuracy = (correct_count / total) * 100
print(f"Final Test Accuracy: {accuracy:.2f}%")
```

```python
import json

config_data = {
    "model_name": "distilbert-base-cased-distilled-squad",
    "task": "question-answering",
    "top_k": 1,
    "framework": "PyTorch / Transformers",
    "evaluation_accuracy": f"{accuracy:.2f}%"
}

with open("model_config.json", "w") as f:
    json.dump(config_data, f, indent=4)

print("Model configurations and metadata successfully saved to model_config.json.")
```
